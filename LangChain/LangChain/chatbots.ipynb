```python
from fastapi import FastAPI, HTTPException
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="Gemma2-9b-It", api_key=GROQ_API_KEY)
model
```

```python
from langchain_core.messages import HumanMessage,SystemMessage,AIMessage
model.invoke([HumanMessage(content="Hii My Name is Rushikesh Cheif AI engineer")])
```

```python
model.invoke(
  [
  HumanMessage(content="Hii My Name is Rushikesh Cheif AI engineer"),
  AIMessage(content="Hello Rushikesh! It's a pleasure to meet you. \n\nThat's an impressive title - Chief AI Engineer! What kind of exciting projects are you working on? \n\nI'm always eager to learn more about the work being done in the field of AI."),
  HumanMessage(content="Hey Whats my name and what do i do?"),
  ]
)
```

```python
## Message History : To make model stateful
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store={}
def get_session_history(session_id:str)->BaseChatMessageHistory:
  if session_id not in store:
    store[session_id]=ChatMessageHistory()
  return store[session_id]
  
with_message_history=RunnableWithMessageHistory(model,get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
resonse=with_message_history.invoke(
  [
    HumanMessage(content="Hii My Name is Rushikesh Cheif AI Engineer"),
  ],
  config=config
  
)
```

```python
resonse.content
```

```python
config1={"configurable":{"session_id":"chat2"}}

resonse=with_message_history.invoke(
  [
    HumanMessage(content="Whats my name?"),
  ],
  config=config1
)
resonse.content
```
