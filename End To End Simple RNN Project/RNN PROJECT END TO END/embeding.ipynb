```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a bad boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
### One hot representation for every word
one_hot_repr=[one_hot(word,voc_size) for word in sent]
one_hot_repr
```

```python
### word Embeding reprsentation

from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
embded_docs
```

```python
## Feature repensentaion
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embded_docs)
```
